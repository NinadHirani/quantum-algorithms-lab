# Shor's Algorithm: A Complete Guide

Shor's algorithm is one of the most famous quantum algorithms that can factor large integers efficiently. This notebook breaks down the core concepts and implementation step by step.

## Setup

First, let's import the necessary libraries.

```python
import numpy as np
from math import gcd
import random
```

---

## Example 1: Understanding GCD (Greatest Common Divisor)

The GCD is fundamental to Shor's algorithm. If we find a random number that shares a factor with N, we can directly find factors.

```python
def find_common_factor(a, N):
    return gcd(a, N)

N = 15
a = 9
common = find_common_factor(a, N)
print(f"GCD({a}, {N}) = {common}")
print(f"If GCD > 1, we found a factor: {N // common}")
```

---

## Example 2: Modular Exponentiation

Computing a^x mod N efficiently is crucial. This is the core operation in finding the period.

```python
def modular_exponent(a, x, N):
    return pow(a, x, N)

a = 2
N = 15
results = []
for x in range(1, 8):
    result = modular_exponent(a, x, N)
    results.append(result)
    print(f"{a}^{x} mod {N} = {result}")

print(f"\nSequence: {results}")
```

---

## Example 3: Finding the Order (Period)

The order r is the smallest positive integer where a^r ≡ 1 (mod N). Finding r efficiently is where quantum computers excel.

```python
def find_order(a, N):
    result = a % N
    order = 1
    while result != 1:
        result = (result * a) % N
        order += 1
        if order > N:
            return None
    return order

a = 7
N = 15
order = find_order(a, N)
print(f"Order of {a} mod {N}: {order}")
print(f"Verification: {a}^{order} mod {N} = {pow(a, order, N)}")
```

---

## Example 4: Extracting Factors from Period

Once we have the period r, we can compute potential factors using: gcd(a^(r/2) ± 1, N)

```python
def extract_factors(a, r, N):
    if r % 2 != 0:
        return None, None
    
    x = pow(a, r // 2, N)
    factor1 = gcd(x + 1, N)
    factor2 = gcd(x - 1, N)
    
    return factor1, factor2

a = 7
r = 4
N = 15

f1, f2 = extract_factors(a, r, N)
print(f"Using a={a}, r={r}, N={N}")
print(f"x = {a}^({r}//2) mod {N} = {pow(a, r // 2, N)}")
print(f"Factor 1: gcd(x+1, N) = {f1}")
print(f"Factor 2: gcd(x-1, N) = {f2}")
print(f"\nVerification: {f1} × {f2} = {f1 * f2}")
```

---

## Example 5: Simple Factorization (N = 15)

Now let's put it all together for a complete factorization.

```python
def shors_algorithm(N, max_attempts=5):
    if N % 2 == 0:
        return 2, N // 2
    
    for attempt in range(max_attempts):
        a = random.randint(2, N - 1)
        
        g = gcd(a, N)
        if 1 < g < N:
            return g, N // g
        
        r = find_order(a, N)
        if r is None or r % 2 != 0:
            continue
        
        f1, f2 = extract_factors(a, r, N)
        if f1 and f2 and 1 < f1 < N and 1 < f2 < N:
            return f1, f2
    
    return None, None

N = 15
factor1, factor2 = shors_algorithm(N)
print(f"Factoring N = {N}")
print(f"Result: {factor1} × {factor2} = {factor1 * factor2}")
```

---

## Example 6: Complete Shor's Algorithm with Verification

Let's factor another number and verify the result step by step.

```python
def shors_with_steps(N, max_attempts=10):
    print(f"Factoring N = {N}\n")
    
    if N % 2 == 0:
        return 2, N // 2
    
    for attempt in range(max_attempts):
        a = random.randint(2, N - 1)
        print(f"Attempt {attempt + 1}: Testing a = {a}")
        
        g = gcd(a, N)
        if 1 < g < N:
            print(f"  ✓ Found via GCD: {g}")
            return g, N // g
        
        r = find_order(a, N)
        if r is None:
            print(f"  ✗ Order not found")
            continue
        
        print(f"  Order r = {r}")
        
        if r % 2 != 0:
            print(f"  ✗ Order is odd")
            continue
        
        f1, f2 = extract_factors(a, r, N)
        if f1 and f2 and 1 < f1 < N and 1 < f2 < N:
            print(f"  ✓ Found factors: {f1} and {f2}")
            return f1, f2
        
        print(f"  ✗ Invalid factors")
    
    return None, None

N = 21
factor1, factor2 = shors_with_steps(N)
print(f"\nFinal Result: {factor1} × {factor2} = {factor1 * factor2}")
print(f"Correct: {factor1 * factor2 == N}")
```

## Summary

Shor's algorithm has two main parts:

1. **Classical**: Reduce factoring problem to period-finding
2. **Quantum**: Use quantum computer to find the period efficiently

In this notebook, we implemented the classical part. The quantum advantage comes from exponentially faster period-finding on a quantum computer.
